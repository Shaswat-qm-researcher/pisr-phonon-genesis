# Material Compound Classifier — Notebook Example

Import `run_notebook` from the classifier and point it at your file.  
All original columns are preserved; a single `Class` column is appended.

## Sheet selection

For Excel workbooks with multiple sheets, pass the sheet name via the `sheet` parameter.  
If omitted, the **first sheet** is loaded automatically.

```python
from material_classifier import run_notebook

df = run_notebook(
    input_path  = 'data/my_dataset.xlsx',   # <-- change me
    output_dir  = 'results',                 # <-- change me
    formula_col = 'formula',                 # or None for auto-detect
    sheet       = 'Sheet1',                  # <-- Excel sheet name; omit or set None for first sheet
    # actual_col  = 'band_gap',              # optional — for performance metrics
    # pred_col    = 'band_gap_pred',
    prefix      = 'run1',
)

df.head(10)
```

## Inspect available sheets first (optional)

If you are unsure of the sheet name, you can list them before calling `run_notebook`:

```python
import pandas as pd

xls = pd.ExcelFile('data/my_dataset.xlsx')  # <-- change me
print('Available sheets:', xls.sheet_names)
```
